```python
from advertisement_generator import render_ad
from IPython.display import display, Image
product = 'PulseFit Protein'
audience = 'Gym Goers'
requirements = 'bold, modern, energetic'
output_path = 'generated_ad.png'
plan = render_ad(product, audience, requirements, output_path)
print('Category:', plan.category)
print('Tagline:', plan.tagline)
print('CTA:', plan.cta)
print('Design-strength score:', str(plan.confidence) + '% (heuristic score)')
display(Image(filename=output_path))
```
